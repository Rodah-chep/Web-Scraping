## Ria Money

```python
url = "https://moneytransguide.com/USA/by-riamoney/"
res = requests.get(url)
soup = BeautifulSoup(res.content, 'html.parser')

ria_df = []

# Getting cities details
city_lists = soup.find_all("ul", class_="hotlinks_countries")
for city_list in city_lists:
    for city in city_list.find_all("li"):
        city_a = city.find("a", class_="href")
        if not city_a:
            continue
        city_name = city_a.get_text(strip=True)
        city_link = f"https://moneytransguide.com{city_a['href']}" if city_a and city_a.has_attr("href") else "No link"
        city_url = city_link
        city_res = requests.get(city_url)
        soup_city = BeautifulSoup(city_res.content, 'html.parser')
        # Extract all details from each city page using tr.gmap_tr
        for tr in soup_city.find_all("tr", class_="gmap_tr"):
            tds = tr.find_all("td")
            row_data = [td.get_text(strip=True) for td in tds]
            div_tag = tds[0].find("div") if tds else None

            if div_tag:
                result = {
                    "city_name": city_name,
                    "city_link": city_link,
                    "link": div_tag.get("link", "No link"),
                    "lat": div_tag.get("lat", "No lat"),
                    "lng": div_tag.get("lng", "No lng"),
                    "g_map": div_tag.get("g_map", "No g_map"),
                    "name": div_tag.get_text(strip=True),
                    "row_data": row_data
                }
                ria_df.append(result)
            else:
                ria_df.append({
                    "city_name": city_name,
                    "city_link": city_link,
                    "link": "No link",
                    "lat": "No lat",
                    "lng": "No lng",
                    "g_map": "No g_map",
                    "name": "No name",
                    "row_data": row_data
                })

ria_df = pd.DataFrame(ria_df)
display(ria_df)
```

```python
url="https://moneytransguide.com/USA/"
response=requests.get(url)
soup=BeautifulSoup(response.content,'html.parser')
print(response.status_code)
titles = soup.find_all("td", class_="anytd_1st")

money_transfer = []  # List to hold all records

for td in titles:
    # Extract alt attribute from img tag
    img_tag = td.find("img")
    alt_text = img_tag["alt"] if img_tag and img_tag.has_attr("alt") else "No alt attribute"

    # Extract href from a tag
    a_tag = td.find("a")
    link = f"https://moneytransguide.com" + a_tag["href"] if a_tag and a_tag.has_attr("href") else "No link"

    ## getting the city details
    url = link
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Failed to fetch page: {url}")
        continue
    soup_city = BeautifulSoup(response.content, 'html.parser')
    city_list = soup_city.find("ul", class_="hotlinks_countries")
    if not city_list:
        continue
    # getting city names and all details
    for city in city_list.find_all("li"):
        city_a = city.find("a", class_="href")
        if not city_a:
            continue
        city_name = city_a.get_text(strip=True)
        city_link = city_a['href']
        city_url = f"https://moneytransguide.com{city_link}"
        city_res = requests.get(city_url)
        if city_res.status_code != 200:
            city_details = "Failed to fetch"
            gmap_data = []
        else:
            soup_city_detail = BeautifulSoup(city_res.content, 'html.parser')
            td_s = soup_city_detail.find("td")
            div_tag = td_s.find("div") if td_s else None
            if div_tag:
                gmap_data = [{
                    "gmap_link": div_tag.get("link", "No link"),
                    "lat": div_tag.get("lat", "No lat"),
                    "lng": div_tag.get("lng", "No lng"),
                    "g_map": div_tag.get("g_map", "No g_map")
                }]
                city_details = ""
            else:
                gmap_data = []
                city_details = ""
        # Append as a dictionary, including gmap_data
        money_transfer.append({
            "alt_text": alt_text,
            "link": link,
            "city_name": city_name,
            "city_link": city_link,
            "city_url": city_url,
            "city_details": city_details,
            "gmap_data": gmap_data
        })

# Flatten gmap_data for DataFrame
records = []
for entry in money_transfer:
    if entry["gmap_data"]:
        for gmap in entry["gmap_data"]:
            record = {
                "alt_text": entry["alt_text"],
                "link": entry["link"],
                "city_name": entry["city_name"],
                "city_link": entry["city_link"],
                "city_url": entry["city_url"],
                "city_details": entry["city_details"],
                "gmap_link": gmap.get("gmap_link", "No link"),
                "lat": gmap.get("lat", "No lat"),
                "lng": gmap.get("lng", "No lng"),
                "g_map": gmap.get("g_map", "No g_map"),
                "address": gmap.get("address", "No address"),
                "name": gmap.get("name", "No name")
            }
            records.append(record)
    else:
        record = {
            "alt_text": entry["alt_text"],
            "link": entry["link"],
            "city_name": entry["city_name"],
            "city_link": entry["city_link"],
            "city_url": entry["city_url"],
            "city_details": entry["city_details"],
            "gmap_link": "No link",
            "lat": "No lat",
            "lng": "No lng",
            "g_map": "No g_map",
            "address": "No address",
            "name": "No name"
        }
        records.append(record)

money_transfer_df = pd.DataFrame(records)
display(money_transfer_df)
```
